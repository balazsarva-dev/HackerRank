```python
import re
from typing import Callable
import logging

logger = logging.getLogger(__name__)
logging.basicConfig(level=logging.DEBUG, format='%(asctime)s - %(levelname)s - %(message)s')

def logging_decorator(func):
    def wrapper(*args, **kwargs):
        logger.info(f"Calling '{func.__name__}' func with args: {args} and kwargs: {kwargs}")
        result = func(*args, **kwargs)
        logger.info(f"Returned: {result}")
        return result
    return wrapper

def get_valid_number(prompt: str, condition: Callable = None) -> int:
    while True:
        try:
            number = int(input(prompt))
            if condition is None or condition(number):
                return number
            else:
                logger.warning("Out of range error! Please try again! ")
        except ValueError:
            logger.error("Invalid input! Please enter a number! ")

@logging_decorator
def perfect_number(n: int) -> bool:
    return sum(i for i in range(1, n//2 + 1) if n % i == 0) == n

@logging_decorator
def valid_password(password: str, min_length: int, max_lenght: int) -> bool:
    if len(password) not in range(min_length, max_lenght + 1):
        return False
    patterns = [r'[0-9]', r'[a-z]', r'[A-Z]', r'[@$!%*?&]']
    for pattern in patterns:
        if not re.search(pattern, password):
            return False
    return True

def main():
    test_iteration_number = get_valid_number("Enter a number: ", lambda x: 1 <= x <= 5)
    for i in range(test_iteration_number):
        logger.info(f"Test #{i + 1}")
        test_number = get_valid_number("Enter a number: ", lambda x: 1 <= x <= 100)
        logger.info(f"Valid number: {test_number}")
        perfect_number(test_number)

if __name__ == "__main__":
    main()
```

# Decorators
Decorators are functions that modify the behavior of another function or method.
They are often used for logging, access control, memoization, and more.

```python
def logging_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function '{func.__name__}' with arguments {args} and {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function '{func.__name__}' returned {result}")
        return result
    return wrapper

@logging_decorator
def add(a, b):
    return a + b

# Usage
add(3, 7)
```

# Generators
Generators are functions that yield values one at a time using the `yield` keyword.
They are memory-efficient and useful for handling large datasets.

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Usage
for num in fibonacci(10):
    print(num)
```

```python
def fib_alter():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

for idx, num in enumerate(fib_alter()):
    if idx == 10:
        break
    print(num)
```

# Advanced Decorators
Decorators can be nested, accept arguments, and be used for complex tasks like timing and logging.

```python
import time

def timing_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"Execution time: {end_time - start_time:.4f} seconds")
        return result
    return wrapper

def logging_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function '{func.__name__}' with arguments {args} and {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function '{func.__name__}' returned {result}")
        return result
    return wrapper

@timing_decorator
@logging_decorator
def compute_sum(n):
    return sum(range(n))

# Usage
compute_sum(1000000)
```

# Advanced Generators
Generators can be used for pipelines and concise expressions.

```python
def read_numbers():
    for i in range(1, 11):
        yield i

def filter_even(numbers):
    for num in numbers:
        if num % 2 == 0:
            yield num

def square(numbers):
    for num in numbers:
        yield num ** 2

# Usage
numbers = read_numbers()
even_numbers = filter_even(numbers)
squared_numbers = square(even_numbers)

for result in squared_numbers:
    print(result)
```

# How Decorators Work
Decorators are higher-order functions that take a function as input, wrap it with additional functionality, and return the modified function.

```python
def simple_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before the function call")
        result = func(*args, **kwargs)
        print("After the function call")
        return result
    return wrapper

def greet(name):
    print(f"Hello, {name}!")

# Manually applying the decorator
decorated_greet = simple_decorator(greet)
decorated_greet("Alice")

# Using @ syntax
@simple_decorator
def farewell(name):
    print(f"Goodbye, {name}!")

farewell("Bob")
```

# Stateful Decorators
Stateful decorators maintain state across multiple calls. This is useful for tracking function calls or caching results.

```python
class CallCounter:
    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f"Call {self.count} to {self.func.__name__}")
        return self.func(*args, **kwargs)

@CallCounter
def greet(name):
    print(f"Hello, {name}!")

# Usage
greet("Alice")
greet("Bob")
greet("Charlie")
```

# Coroutines
Coroutines are a special type of generator that can consume values using the `send()` method. They are useful for pipelines and event-driven programming.

```python
def averager():
    total = 0
    count = 0
    average = None
    while True:
        num = yield average
        total += num
        count += 1
        average = total / count

# Usage
avg = averager()
next(avg)  # Prime the coroutine
print(avg.send(10))  # Output: 10.0
print(avg.send(20))  # Output: 15.0
print(avg.send(30))  # Output: 20.0
```

# Advanced SQL: Joins and Transactions
Learn how to use joins to combine data from multiple tables and transactions to ensure data consistency.

```python
import sqlite3

connection = sqlite3.connect("example.db")
cursor = connection.cursor()

# Create tables
cursor.execute("""
CREATE TABLE IF NOT EXISTS guests (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    age INTEGER NOT NULL
)
""")
cursor.execute("""
CREATE TABLE IF NOT EXISTS reservations (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    guest_id INTEGER NOT NULL,
    room_number INTEGER NOT NULL,
    FOREIGN KEY (guest_id) REFERENCES guests (id)
)
""")

# Insert data
cursor.execute("INSERT INTO guests (name, age) VALUES (?, ?)", ("Alice", 25))
cursor.execute("INSERT INTO reservations (guest_id, room_number) VALUES (?, ?)", (1, 101))
connection.commit()

# Perform an inner join
cursor.execute("""
SELECT guests.name, reservations.room_number
FROM guests
INNER JOIN reservations ON guests.id = reservations.guest_id
""")
rows = cursor.fetchall()
for row in rows:
    print(row)

connection.close()
```

# Advanced pymysql: Transactions and Error Handling
Learn how to use transactions and handle errors gracefully with `pymysql`.

```python
import pymysql

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

try:
    # Start a transaction
    connection.begin()

    # Insert data
    cursor.execute("INSERT INTO guests (name, age) VALUES (%s, %s)", ("Grace", 35))
    cursor.execute("INSERT INTO reservations (guest_id, room_number) VALUES (%s, %s)", (3, 201))

    # Commit the transaction
    connection.commit()
    print("Transaction committed successfully!")
except Exception as e:
    # Rollback the transaction in case of an error
    connection.rollback()
    print(f"Transaction failed: {e}")
finally:
    connection.close()
```

# Advanced aiomysql: Transactions and Connection Pooling
Learn how to perform asynchronous transactions and manage connections efficiently with `aiomysql`.

```python
import asyncio
import aiomysql

async def main():
    connection = await aiomysql.connect(
        host="localhost",
        user="root",
        password="password",
        db="test_db"
    )
    async with connection.cursor() as cursor:
        try:
            # Start a transaction
            await connection.begin()

            # Insert data
            await cursor.execute("INSERT INTO guests (name, age) VALUES (%s, %s)", ("Hannah", 28))
            await cursor.execute("INSERT INTO reservations (guest_id, room_number) VALUES (%s, %s)", (4, 202))

            # Commit the transaction
            await connection.commit()
            print("Transaction committed successfully!")
        except Exception as e:
            # Rollback the transaction in case of an error
            await connection.rollback()
            print(f"Transaction failed: {e}")
        finally:
            connection.close()

asyncio.run(main())
```

# Corrected Indexing Example
This example demonstrates the performance difference between queries with and without an index.

```python
import pymysql
import time

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Create an index on the 'name' column
cursor.execute("CREATE INDEX idx_guest_name ON guests(name)")

# Query without index (drop the index first to simulate)
cursor.execute("DROP INDEX idx_guest_name ON guests")
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query without index took: {time.time() - start_time:.4f} seconds")

# Recreate the index
cursor.execute("CREATE INDEX idx_guest_name ON guests(name)")

# Query with index
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query with index took: {time.time() - start_time:.4f} seconds")

connection.close()
```
