```python
def recursive_max(lst):
    print(f"Hívás: {lst}")
    if len(lst) == 1:
        print(f"Base case: {lst[0]}")
        return lst[0]
    else:
        sub_max = recursive_max(lst[1:])
        print(f"Összehasonlítás: {lst[0]} vs {sub_max}")
        return lst[0] if lst[0] > sub_max else sub_max

print("Maximum:", recursive_max([5, 8, 3]))
```

```python
def rec_fact(n):
    if n == 1:
        return 1
    else:
        result = rec_fact(n - 1)
        return n * result if n > 1 else 1
```

```python
rec_fact(5)  # Example call with a limit
```

```python
def rec_sum(some_list):
    return some_list[0] + rec_sum(some_list[1:]) if some_list else 0
```

```python
rec_sum([1, 2, 3, 4, 5])
```

```python
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
```

```python
result = fib()
for idx, num in enumerate(result):
    if idx == 10:
        break
    print(num, end=' ')
```
