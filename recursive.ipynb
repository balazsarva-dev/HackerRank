```python
class Valid:
    def set_valid_string(self, prompt: str, *args):
        while True:
            user_input = input(prompt)
            if user_input in args:
                self.valid = user_input
                return
    def get_valid_string(self):
        return self.valid

class Process:
    number_of_process = 0

    def __init__(self):
        self.valid = Valid()
        Process.number_of_process += 1

    def want_to_play(self):
        self.valid.set_valid_string("Do you want to play? (yes/no): ", "yes", "no")
        return self.valid.get_valid_string() == "yes"
    
    def starting_game(self):
        if not self.want_to_play():
            print("Maybe next time. Goodbye!")
            return
        # Game logic would go here
        print("Starting the game...")

game_one = Process()
game_one.starting_game()
```

```python
def recursive_max(lst):
    print(f"Hívás: {lst}")
    if len(lst) == 1:
        print(f"Base case: {lst[0]}")
        return lst[0]
    else:
        sub_max = recursive_max(lst[1:])
        print(f"Összehasonlítás: {lst[0]} vs {sub_max}")
        return lst[0] if lst[0] > sub_max else sub_max

print("Maximum:", recursive_max([5, 8, 3]))
```

```python
def rec_fact(n):
    if n == 1:
        return 1
    else:
        result = rec_fact(n - 1)
        return n * result if n > 1 else 1
```

```python
rec_fact(5)  # Example call with a limit
```

```python
def rec_sum(some_list):
    return some_list[0] + rec_sum(some_list[1:]) if some_list else 0
```

```python
rec_sum([1, 2, 3, 4, 5])
```

```python
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
```

```python
result = fib()
for idx, num in enumerate(result):
    if idx == 10:
        break
    print(num, end=' ')
```
