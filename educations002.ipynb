# Stored Procedures
Learn how to create and use stored procedures to encapsulate logic in the database.

```python
DELIMITER $$

CREATE PROCEDURE GetGuestReservations(IN guest_name VARCHAR(255))
BEGIN
    SELECT g.name, r.room_number
    FROM guests g
    INNER JOIN reservations r ON g.id = r.guest_id
    WHERE g.name = guest_name;
END $$

DELIMITER ;
```

```python
import pymysql

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Call the stored procedure
guest_name = "Alice"
cursor.callproc("GetGuestReservations", [guest_name])

# Fetch results
rows = cursor.fetchall()
for row in rows:
    print(row)

connection.close()
```

# Indexing
Learn how to create indexes to improve query performance.

```python
CREATE INDEX idx_guest_name ON guests(name);
```

```python
import pymysql
import time

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Query without index
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query without index took: {time.time() - start_time:.4f} seconds")

# Query with index
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query with index took: {time.time() - start_time:.4f} seconds")

connection.close()
```

# Query Optimization
Learn how to analyze and optimize queries using EXPLAIN.

```python
EXPLAIN SELECT g.name, r.room_number
FROM guests g
INNER JOIN reservations r ON g.id = r.guest_id
WHERE g.name = 'Alice';
```

```python
import pymysql

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Use EXPLAIN to analyze the query
cursor.execute("""
EXPLAIN SELECT g.name, r.room_number
FROM guests g
INNER JOIN reservations r ON g.id = r.guest_id
WHERE g.name = 'Alice'
""")
rows = cursor.fetchall()
for row in rows:
    print(row)

connection.close()
```
