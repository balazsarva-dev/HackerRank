# Stored Procedures
Learn how to create and use stored procedures to encapsulate logic in the database.

```python
DELIMITER $$

CREATE PROCEDURE GetGuestReservations(IN guest_name VARCHAR(255))
BEGIN
    SELECT g.name, r.room_number
    FROM guests g
    INNER JOIN reservations r ON g.id = r.guest_id
    WHERE g.name = guest_name;
END $$

DELIMITER ;
```

```python
import pymysql

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Call the stored procedure
guest_name = "Alice"
cursor.callproc("GetGuestReservations", [guest_name])

# Fetch results
rows = cursor.fetchall()
for row in rows:
    print(row)

connection.close()
```

```python
import random

def generate_random_list() -> list[int]:
    return [random.randint(1, 100) for _ in range(10)]

def get_max_index_in_list(some_list: list[int]) -> int:
    max_idx = 0
    for i in range(len(some_list)):
        if some_list[i] > some_list[max_idx]:
            max_idx = i
    return max_idx

def get_max_index_in_list_alter(some_list: list[int]) -> int:
    return some_list.index(max(some_list))

some_list = generate_random_list()
test = get_max_index_in_list(some_list)
print(f"List: {some_list}")
print(f"Max index: {test}")
print(f"Max value: {some_list[test]}")
check_test = get_max_index_in_list_alter(some_list)
print(f"Max index (alter): {check_test}")
```

```python
from collections import defaultdict
from pprint import pprint

def get_pizzas() -> dict:
    return {
                'pepperoni': 1.99,
                'mushroom': 2.99,
                'sausage': 3.99,
                'onion': 1.99,
                'cheese': 4.99,
                'chicken': 5.99,
                'veggie': 2.99,
                'bbq_chicken': 6.99,
                'hawaiian': 4.99,
            }

def get_prices(items: dict) -> dict:
    prices = defaultdict(tuple)
    for k in items:
        prices[items[k]] += (k,)
    return prices

def count_unique_prices(prices: dict) -> None:
    for k in prices:
        pprint(f"For {k}$ you can choose from {len(prices[k])} pizzas")

default_pizzas = get_pizzas()
original_prices = get_prices(default_pizzas)
pprint(original_prices)
count_unique_prices(original_prices)
```

```python
import pymysql
import time

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Query without index
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query without index took: {time.time() - start_time:.4f} seconds")

# Query with index
start_time = time.time()
cursor.execute("SELECT * FROM guests WHERE name = 'Alice'")
rows = cursor.fetchall()
print(f"Query with index took: {time.time() - start_time:.4f} seconds")

connection.close()
```

# Query Optimization
Learn how to analyze and optimize queries using EXPLAIN.

```python
EXPLAIN SELECT g.name, r.room_number
FROM guests g
INNER JOIN reservations r ON g.id = r.guest_id
WHERE g.name = 'Alice';
```

```python
import pymysql

connection = pymysql.connect(
    host="localhost",
    user="root",
    password="password",
    database="test_db"
)
cursor = connection.cursor()

# Use EXPLAIN to analyze the query
cursor.execute("""
EXPLAIN SELECT g.name, r.room_number
FROM guests g
INNER JOIN reservations r ON g.id = r.guest_id
WHERE g.name = 'Alice'
""")
rows = cursor.fetchall()
for row in rows:
    print(row)

connection.close()
```
