```python
def validate_weight():
    while True:
        weight = input('Enter the weight of the package in pounds: ')
        try:
            weight = float(weight)
            if weight > 0:
                return weight
            print('Please enter a valid weight (must be positive).')
        except ValueError:
            print('Please enter a valid number for weight.')
```

```python
import random

def generate_random_list() -> list[int]:
    return [random.randint(1, 100) for _ in range(10)]

def get_max_index_in_list(some_list: list[int]) -> int:
    max_idx = 0
    for i in range(len(some_list)):
        if some_list[i] > some_list[max_idx]:
            max_idx = i
    return max_idx

def get_max_index_in_list_alter(some_list: list[int]) -> int:
    return some_list.index(max(some_list))

some_list = generate_random_list()
test = get_max_index_in_list(some_list)
print(f"List: {some_list}")
print(f"Max index: {test}")
print(f"Max value: {some_list[test]}")
check_test = get_max_index_in_list_alter(some_list)
print(f"Max index (alter): {check_test}")
```

```python
from collections import defaultdict
from pprint import pprint

def get_pizzas() -> dict:
    return {
                'pepperoni': 1.99,
                'mushroom': 2.99,
                'sausage': 3.99,
                'onion': 1.99,
                'cheese': 4.99,
                'chicken': 5.99,
                'veggie': 2.99,
                'bbq_chicken': 6.99,
                'hawaiian': 4.99,
            }

def get_prices(items: dict) -> dict:
    prices = defaultdict(tuple)
    for k in items:
        prices[items[k]] += (k,)
    return prices

def count_unique_prices(prices: dict) -> None:
    for k in prices:
        pprint(f"For {k}$ you can choose from {len(prices[k])} pizzas")

default_pizzas = get_pizzas()
original_prices = get_prices(default_pizzas)
pprint(original_prices)
count_unique_prices(original_prices)
```
